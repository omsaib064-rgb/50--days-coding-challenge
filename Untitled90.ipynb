```python
1. Calculate Average
```

```python
def calculate_average(values):
    return sum(values) / len(values) if values else 0
```

```python
2. Top Performers
```

```python
def find_top_performers(data, column, threshold):
    return [student for student in data if student[column] >= threshold]
```

```python
3. At Risk Students
```

```python
def find_at_risk_students(data, column, threshold):
    return [student for student in data if student[column] < threshold]
```

```python
4. Batch Summary
```

```python
def batch_summary(data):
    batches = {}

    for student in data:
        b = student["batch"]
        if b not in batches:
            batches[b] = {
                "consumption": [],
                "live": [],
                "assignment_sub": []
            }

        batches[b]["consumption"].append(student["consumption"])
        batches[b]["live"].append(student["live"])
        batches[b]["assignment_sub"].append(student["assignment_sub"])

    summary = {}

    for b in batches:
        summary[b] = {
            "avg_consumption": round(calculate_average(batches[b]["consumption"]), 1),
            "avg_attendance": round(calculate_average(batches[b]["live"]), 1),
            "avg_assignment": round(calculate_average(batches[b]["assignment_sub"]), 1)
        }

    return summary
```

```python
5. Rank Batches
```

```python
def rank_batches(summary):
    best_batch = None
    best_score = 0

    for b, metrics in summary.items():
        score = (metrics["avg_consumption"] + metrics["avg_attendance"]) / 2
        if score > best_score:
            best_score = score
            best_batch = b

    return best_batch
```
