```python
# Insurance Premium Calculator

def calculate_premium(age, policy_type, claims):

    # Base premium dictionary
    premiums = {
        "health": 8000,
        "vehicle": 12000,
        "life": 10000
    }

    # Validate policy type
    if policy_type not in premiums:
        return None

    # Get base premium
    premium = premiums[policy_type]

    # Age adjustment
    if age < 25:
        premium += premium * 0.20

    elif age > 45:
        premium += premium * 0.15

    # Claims adjustment
    if claims > 2:
        premium += premium * 0.25

    elif claims == 0:
        premium -= premium * 0.10

    return premium


# User Inputs
name = input("Enter Customer Name: ")
age = int(input("Enter Age: "))
policy_type = input("Enter Policy Type (health/vehicle/life): ").lower()
claims = int(input("Enter Number of Claims: "))

# Calculate premium
final_premium = calculate_premium(age, policy_type, claims)

# Output
if final_premium is None:
    print("Invalid policy type entered!")

else:
    base_prices = {
        "health": 8000,
        "vehicle": 12000,
        "life": 10000
    }

    print("\n===== Insurance Premium Details =====")
    print("Customer Name :", Ravi)
    print("Policy Type   :", Health)
    print("Base Premium  : ₹", 8000)
    print("Final Premium : ₹", int(8640))
```
