```python
# Total price before discount
total_price = product_price * quantity

print("Total Price before discount:", total_price)
```

```python
# Calculate discount amount
discount_amount = total_price * discount

# Price after discount
price_after_discount = total_price - discount_amount

print("Discount Amount:", discount_amount)
print("Price after discount:", price_after_discount)
```

```python
# GST percentage
gst_rate = 0.18

# Calculate GST amount
gst_amount = price_after_discount * gst_rate

# Final bill amount
final_bill = price_after_discount + gst_amount

print("GST Amount:", gst_amount)
print("Final Bill Amount:", final_bill)
```

```python
print("\n----- BILL SUMMARY -----")
print(f"Order ID: {order_id}")
print(f"Customer: {customer_name}")
print(f"Total Price: ₹{total_price:.2f}")
print(f"Discount Applied: {'Yes' if discount > 0 else 'No'}")
print(f"Final Amount (including GST): ₹{final_bill:.2f}")
print("------------------------")
```
