```python
rides = (
    (101, "Ravi", "Bangalore", 250, 4.5),
    (102, "Anjali", "Hyderabad", 180, 4.2),
    (103, "Kiran", "Bangalore", 300, 4.8),
    (104, "Meena", "Chennai", 220, 4.0),
    (105, "Arjun", "Bangalore", 150, 3.9)
)

total_rides = len(rides)
print("Total Rides:", total_rides)
```

```python
rides = (
    (101, "Ravi", "Bangalore", 250, 4.5),
    (102, "Anjali", "Hyderabad", 180, 4.2),
    (103, "Kiran", "Bangalore", 300, 4.8),
    (104, "Meena", "Chennai", 220, 4.0),
    (105, "Arjun", "Bangalore", 150, 3.9)
)

bangalore_rides = [ride for ride in rides if ride[2] == "Bangalore"]

print("Bangalore Rides:")
for ride in bangalore_rides:
    print(ride)
```

```python
loop Method
```

```python
for ride in rides:
    if ride[2] == "Bangalore":
        print(ride)
```

```python
Total Revenue from Bangalore
```

```python

total_revenue = sum(ride[3] for ride in rides if ride[2] == "Bangalore")

print("Total Revenue from Bangalore:", total_revenue)
```

```python
Find Highest Rated Driver
```

```python
highest_rated = max(rides, key=lambda ride: ride[4])

print("Highest Rated Driver:", highest_rated)
```

```python
2: Calculate average
```

```python


# Step 1: Group ratings by city
city_ratings = {}

for ride in rides:
    city = ride[2]
    rating = ride[4]

    if city not in city_ratings:
        city_ratings[city] = []

    city_ratings[city].append(rating)

# Step 2: Calculate average
avg_ratings = {}

for city, ratings in city_ratings.items():
    avg_ratings[city] = round(sum(ratings) / len(ratings), 1)

print(avg_ratings)
```
