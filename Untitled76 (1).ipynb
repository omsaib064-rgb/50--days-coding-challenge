```python
: Find Longest Student Name
```

```python
students = ("Anu", "Ravikumar", "Jo")

longest_name = max(students, key=len)

print("Longest name:", longest_name)
```

```python
: Join Subjects into Single String 
```

```python
subjects = ("Math", "Science", "English")

result = ", ".join(subjects)

print("Subjects:", result)
```

```python
3: Reverse Each Student Name
```

```python
students = ("Anu", "Ravi")

reversed_names = tuple(name[::-1] for name in students)

print("Reversed names:", reversed_names)
```

```python
4: Count Vowels in Each Name
```

```python
students = ("Anu", "Ravi")

vowels = "aeiouAEIOU"

vowel_count = {name: sum(1 for char in name if char in vowels) for name in students}

print("Vowel count:", vowel_count)
```

```python
5: Convert Tuple to Uppercase String
```

```python
courses = ("python", "sql", "power bi")

result = " | ".join(course.upper() for course in courses)

print("Courses:", result)
```
