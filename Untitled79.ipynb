```python
. Calculate total & average for each student 
```

```python
students = {
    "Asha": {"Math": 78, "Science": 85, "English": 67},
    "Ravi": {"Math": 35, "Science": 40, "English": 50},
    "John": {"Math": 90, "Science": 92, "English": 88}
}

for name, marks in students.items():
    total = sum(marks.values())
    average = total / len(marks)
    
    print(name, "→ Total:", total, "Average:", round(average, 2))
```

```python
2. Identify top scorer 
```

```python
students = {
    "Asha": {"Math": 78, "Science": 85, "English": 67},
    "Ravi": {"Math": 35, "Science": 40, "English": 50},
    "John": {"Math": 90, "Science": 92, "English": 88}
}

# Calculate total marks and find top scorer
top_scorer = max(students, key=lambda name: sum(students[name].values()))

print("Top Scorer:", top_scorer)
```

```python
. Find students who failed in any subject 
```

```python
students = {
    "Asha": {"Math": 78, "Science": 85, "English": 67},
    "Ravi": {"Math": 35, "Science": 40, "English": 50},
    "John": {"Math": 90, "Science": 92, "English": 88}
}

failed_students = []

for name, marks in students.items():
    if any(score < 40 for score in marks.values()):
        failed_students.append(name)

print("Failed Students:", failed_students)
```
