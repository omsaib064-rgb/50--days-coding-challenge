```python
1️⃣ Remove Duplicates & Preserve Order

Concepts: set, list, loop
```

```python
ids = [101, 102, 101, 103, 102, 104]

unique_ids = []
seen = set()

for id in ids:
    if id not in seen:
        unique_ids.append(id)
        seen.add(id)

print(unique_ids)
```

```python
2️⃣ Frequency Counter (Real-time Analytics)

Concepts: dictionary, get, loop
```

```python
products = ["apple", "banana", "apple", "orange", "banana", "apple"]

freq = {}

for item in products:
    freq[item] = freq.get(item, 0) + 1

print(freq)
```

```python
3️⃣ Find Top Scorer

Concepts: max, dictionary
```

```python
marks = {"A": 85, "B": 92, "C": 88}

topper = max(marks, key=marks.get)

print(topper)
```

```python
4️⃣ Common Elements in Two Lists

Concepts: set, intersection
```

```python
skills1 = ["Python", "SQL", "Excel"]
skills2 = ["SQL", "Power BI", "Python"]

common = list(set(skills1).intersection(set(skills2)))

print(common)
```

```python
5️⃣ Sort Dictionary by Values

Concepts: sorted, lambda
```

```python
prices = {"apple": 50, "banana": 20, "orange": 30}

sorted_prices = sorted(prices.items(), key=lambda x: x[1])

print(sorted_prices)
```
