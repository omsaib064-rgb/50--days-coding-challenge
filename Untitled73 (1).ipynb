```python
data = "Ride101-Ramesh-Late Arrival-Bangalore, Ride102-Suresh-rude behavior-Hyderabad, Ride103-Mahesh-Late arrival-Bangalore"

# Task 1: Clean Data
clean_data = data.lower()

# Task 2: Count total complaints
records = clean_data.split(",")
total_complaints = len(records)

# Task 3: Count late arrival
late_count = clean_data.count("late arrival")

# Task 4: Count Bangalore complaints
bangalore_count = clean_data.count("bangalore")

# Task 5: Extract first driver name
first_driver = records[0].split("-")[1]

# Task 6: Replace issue names
standard_data = clean_data.replace("rude behavior", "rude_behavior")
standard_data = standard_data.replace("late arrival", "late_arrival")

# Output
print("Total Complaints:", total_complaints)
print("Late Arrival Count:", late_count)
print("Bangalore Count:", bangalore_count)
print("First Driver:", first_driver)
print("Standardized Data:", standard_data)
```
