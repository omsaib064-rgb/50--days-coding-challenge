```python
1. STRING Challenge
Problem: Longest Substring Without Repeating Characters
Description: Given a string, find the length of the longest substring without repeating characters.
Input:
"abcabcbb"
```

```python
def longest_substring(s):
    char_set = set()
    left = 0
    max_length = 0

    for right in range(len(s)):
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1

        char_set.add(s[right])
        max_length = max(max_length, right - left + 1)

    return max_length

# Input
print(longest_substring("abcabcbb"))  # Output: 3
```

```python
2. LIST Challenge
Problem: Remove Duplicates While Preserving Order
Description: Given a list of integers, remove duplicates while maintaining the original order.
Input:
[1, 2, 2, 3, 4, 3, 5]
```

```python
def remove_duplicates(lst):
    seen = set()
    result = []

    for num in lst:
        if num not in seen:
            seen.add(num)
            result.append(num)

    return result

# Input
print(remove_duplicates([1, 2, 2, 3, 4, 3, 5]))
# Output: [1, 2, 3, 4, 5]
```

```python
3. TUPLE Challenge
Problem: Find Maximum and Minimum
Description: Given a tuple of numbers, find the maximum and minimum values.
Input:
(10, 5, 20, 8)

```

```python
def find_min_max(tpl):
    return min(tpl), max(tpl)

# Input
print(find_min_max((10, 5, 20, 8)))
# Output: (5, 20)
```

```python
4. DICTIONARY Challenge
Problem: Character Frequency Counter
Description: Given a string, count the frequency of each character and store it in a dictionary.
Input:
"programming"
Challenge (Intermediate)
Problem: Group Words by Length
Description: Given a list of words, group them based on their length using a dictionary.
Input:
["cat", "dog", "elephant", "bat", "tiger"]
```

```python
def char_frequency(s):
    freq = {}

    for char in s:
        freq[char] = freq.get(char, 0) + 1

    return freq

# Input
print(char_frequency("programming"))
```
